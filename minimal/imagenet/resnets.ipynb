basic idea is residual connections. (but done in extreme). The idea is that if extra layers aren't useful the layers can just learn 0 and pass through so gradients can learn very deep layers without degrading

```python
import torch.nn as nn
import torch.nn.functional as F

class Block(nn.Module):
    def __init__(self, c):
        super().__init__()
        self.f = nn.Sequential(
            nn.Conv2d(c, c, 3, padding=1), nn.BatchNorm2d(c), nn.ReLU(),
            nn.Conv2d(c, c, 3, padding=1), nn.BatchNorm2d(c))
    def forward(self, x):
        return F.relu(self.f(x) + x)
```

x → conv3×3 → BN → ReLU → conv3×3 → BN → (+ x) → ReLU
